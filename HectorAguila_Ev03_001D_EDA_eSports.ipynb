- **Predecir el rendimiento**: Usar regresión para entender qué variables disparan el `performance_score`.
- **Detectar talento**: Usar clasificación para ver quién tiene perfil de MVP.

### 1.2. Sobre el Dataset

Asumo que al ser el dataset oficial con el que vamos a trabajar durante el semestre, la estructura es fija. Sin embargo, para que un modelo de este tipo sea realmente útil en la vida real, considero que debería ser complementado con los siguientes features:

- **Carga de entrenamiento**: Para entender si la fatiga es por exceso de horas o por la intensidad del torneo.

Menciono esto, entendiendo medianamente el contexto competitivo, donde estos datos marcan  la diferencia.

## 2. Fase: Comprensión de los Datos

### 2.1. Clasificación de Variables

Primero separé las variables (Nominal, Ordinal, Continuo, Discreto). Esto me permitió decidir, que los Roles no tienen un orden (Nominal) pero las instancias del torneo sí (Ordinal).

### 2.2. El problema del match_outcome

Acá encontré un "ruido" importante: casi el 100% de los registros son victorias.

- Descarté esta variable para entrenar modelos de clasificación. Si el modelo solo ve victorias, va a aprender que "siempre se gana", y eso no sirve para predecir nada realmente.

---

## 3. Fase: Preparación de Datos

### 3.1. Limpieza de Outliers: El límite humano

Para la limpieza, según investigación, fundamenté mi decisión en: "el límitie humano", según la **IAAF (World Athletics)**

- **Filtro de 120ms**: Borré cualquier registro con un tiempo de respuesta menor a eso. Porque según la neurofisiología, un humano no puede reaccionar tan rápido. Esos datos son errores de captura o ruido, y si los dejo, el modelo se "rompe".
- **IQR**: Para el resto de los valores extremos (los que son muy lentos), use la limpieza de las colas para quedarme con el corazón de los datos.

### 3.2. Selección de Características

Saqué el `player_id` ya que lo consideré un dato administrativo que solo genera ruido. El modelo tiene que aprender a reconocer un buen jugador por su precisión y sus kills, no por su número de documento. Entonces, el modelo debe aprender **qué hacen** los jugadores, no **cómo se llaman**.

### 3.3. Transformaciones: Estandarización

Para estandarizar usé `StandardScaler`. Porque hay variables en milisegundos y otras en unidades de kills. Sin estandarizar, el modelo le daría más importancia a los números grandes solo por sesgo.

### 2.1. Clasificación de Variables

Primero separé las variables (Nominal, Ordinal, Continuo, Discreto). Esto permitió observar que los Roles no tienen un orden (Nominal) pero las instancias del torneo sí (Ordinal).


### 2.2. Estadística Descriptiva
Se visualizan las estadísticas del dataset para un análisis exploratorio descriptivo.


```python
df.describe().T
```

**min**: Al analizar las estadísticas, se ha podido detectar un posible outlier o error en la variable **reaction_time_ms**. Al ver esta variable, se pudo notar que no tenía sentido (muy bajo). Ese dato podría hacer creer al modelo que un player reacciona a **118ms**, lo que es muy difícil biológicamente. Se puso un filtro de 120ms, para poder asegurar la calidad del dato.

**std**: Mientras el tiempo de reacción se mueve en cientos (**reaction_time_ms: 324.11**), las bajas se mueven en decenas (**kilss: 30**). La diferencia de magnitud hace obligatoria la estandarización con **Standardscaler** para queno hayan sesgos por el tamaño de números.


### 2.3. Matriz de Correlación (Heatmap)
Para finalizar la etapa de comprensión de los datos, generamos una matriz de correlación de Pearson. Esta herramienta visual nos permite identificar el grado de relación entre las variables numéricas.


```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlación de Pearson')
plt.show()
```

Al ver la matriz, se puede notar que las kills y el accuracy tienen una correlación positiva con el performance score. Esto es lógico, ya que mientras más bajas y precisión tenga el player, mejor será su puntaje. Por otro lado, las muertes tienen una correlación negativa, lo que indica que a mayor cantidad de muertes, menor será el puntaje de rendimiento.


### 2.4. Identificación Visual de Outliers
Post análisis descriptivo inicial y con la identificación de ciertos valores atípicos, se procede a realizar una identificación visual de estos mediante un diagrama de caja (**Boxplot**).


```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['reaction_time_ms'], color='lightcoral')
plt.title('Distribución de Reaction Time (Original)')
plt.xlabel('Milisegundos (ms)')
plt.show()
```

El diagrama confirma la presencia de valores atípicos, representados por los puntos que se sitúan fuera de los 'bigotes' del gráfico. Mientras que la caja central concentra el grueso de los tiempos de reacción normales, esos puntos aislados representan las anomalías que detectamos previamente. Esto valida visualmente nuestra decisión de aplicar un proceso de limpieza.


### 3.1. Curaduría y Limpieza Técnica
Tras validar la presencia de anomalías, procedemos con la limpieza técnica del dataset.


```python
# Eliminación de variables administrativas
df_prep = df.drop(columns=['record_id', 'player_id'])
Q1 = df_prep['reaction_time_ms'].quantile(0.25)
Q3 = df_prep['reaction_time_ms'].quantile(0.75)
IQR = Q3 - Q1
lim_sup = Q3 + 1.5 * IQR
df_clean = df_prep.query('reaction_time_ms >= 120 and reaction_time_ms <= @lim_sup').copy()
print(f"Limpieza completada. Registros limpios: {len(df_clean)}")
```

El código aplica la metodología de Rango Intercuartílico (IQR), técnica estándar para la detección de outliers. Calculamos la distancia entre el primer y tercer cuartil para definir un límite superior lógico, y lo combinamos con un filtro biológico de 120ms. Esta lógica nos permite 'podar' los extremos de la distribution de forma equilibrada y objetiva.


### 3.2. Verificación de Limpieza
Se realiza una validación visual post-limpieza.


```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=df_clean['reaction_time_ms'], color='lightgreen')
plt.title('Distribución de Reaction Time (Limpia)')
plt.xlabel('Milisegundos (ms)')
plt.show()
```

Como se puede observar en la comparativa, hemos logrado eliminar los ruidos instrumentales y los valores biológicamente improbables, centrando la muestra en el comportamiento real de los jugadores. Con este dataset depurado, garantizamos que el modelo aprenda patrones de rendimiento válidos, evitando sesgos provocados por datos anómalos.


### 3.3. Transformación y Estandarización
Fnalmente, con el dataset limpio y depurado, se procede a la última fasee de preparación.


```python
# Encoding
df_encoded = pd.get_dummies(df_clean, columns=['player_role', 'map_played', 'team_name'], drop_first=True)
match_map = {'Group Stage': 1, 'Qualifier': 2, 'Playoff': 3, 'Semi-Final': 4, 'Final': 5}
df_encoded['match_type_numeric'] = df_encoded['match_type'].map(match_map)
df_final = df_encoded.drop(columns=['match_type', 'match_outcome'])

# Scaling
scaler = StandardScaler()
cols_to_scale = ['kills', 'assists', 'deaths', 'accuracy_percent', 'reaction_time_ms', 'fatigue_index']
df_final[cols_to_scale] = scaler.fit_transform(df_final[cols_to_scale])
df_final.head().round(3).T
```

La salida final presenta el dataset completamente transformado y listo para el modelamiento. Se puede observar que las variables categóricas han sido convertidas a un formato binario (One-Hot Encoding), mientras que la jerarquía de las instancias del torneo se refleja ahora en valores numéricos del 1 al 5. Asimismo, las métricas de rendimiento han sido estandarizadas, lo que permite que todas las variables contribuyan equitativamente al modelo, independientemente de su unidad original. Este formato final garantiza la máxima eficiencia algorítmica y previene sesgos provocados por las diferencias de magnitud.
